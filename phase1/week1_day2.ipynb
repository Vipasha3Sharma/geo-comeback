```python
row = {"district": "South Delhi", "cases": 1520, "area_km2": 250.0}

print(row)
print(row["district"])
print(row["cases"])
```

```python
row["cases"] = 1600              # change an existing value
row["year"] = 2022               # add a new key (a new "column")

print(row)
print(row.keys())                # all the keys
print(row.values())              # all the values
print("cases" in row)            # does this key exist?
print(row.get("population"))     # safe lookup: gives None if missing
```

```python
table = [
    {"district": "South Delhi", "cases": 1520},
    {"district": "North Delhi", "cases": 980},
    {"district": "East Delhi",  "cases": 1210},
]

print(len(table))              # number of rows
print(table[1])                # the 2nd row (a dictionary)
print(table[1]["district"])    # the 2nd row's district
```

```python
cities = [
    {"name" : "Bengaluru" , "lat" : 12.97, "long" : 77.58},
    {"name" : "Delhi" , "lat" : 28.61, "long" : 77.23},
    {"name" : "Mumbai" , "lat" : 19.07, "long" : 72.87}
]
print(cities[-1]["name"])
```

```python
import pandas as pd
import geopandas as gpd

df = pd.DataFrame(cities)
gdf = gpd.GeoDataFrame(df, geometry = gpd.points_from_xy(df.long,df.lat), crs="EPSG:4326")
print(gdf)
gdf.plot(color = "red")
```
